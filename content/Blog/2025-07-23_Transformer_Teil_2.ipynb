## Positions-Encoding

Bisher haben wir unter Anderem gesehen, wie Multi-Head Attention ermöglicht, das Wörter sich auf andere Wörter im Satz konzentrieren können, um einen reichhaltigen Kontextvektor zu bilden. Ein Problem dabei ist allerdings, dass die Aufmerksamkeit nicht positionsagnostisch ist. Das bedeutet, ein Wort hat keine inhärente Information darüber, ob es das erste, zweite oder x-te Wort im Satz ist, oder ob ein Wort direkt daneben oder am anderen Ende des Satzes steht.

Das folgende Beispiel sollte es veranschaulichen.

1. _Hund beißt Mann._
2. _Mann beißt Hund._

Die Wort-Embeddings für `Hund`, `beißen` und `Mann` wären in beiden Sätzen identisch. Ohne Positionsinformationen würde der Attention-Mechanismus im Grunde dieselben Aufmerksamkeitspunkte berechnen, da die Wörter die gleichen sind, nur die Reihenfolge ist eine andere andere. Wir sind uns aber sicher auch einig, dass die Bedeutung absolut unterschiedlich ist.

Die Transformatoren von denen wir reden haben im Gegensatz zu rekurrenten Netzwerken (wie RNNs oder LSTMs) keine explizite rekursive Struktur, die die Reihenfolge der Elemente verarbeitet. Daher benötigen sie einen Mechanismus, um die positionelle Information der Wörter in die Eingabe-Embeddings zu integrieren.

An der Stelle kommt das sogenannte **Positions-Encoding** ins Spiel.

Die Kernidee ist, jedem Wort-Embedding einen Vektor hinzuzufügen, der seine Position im Satz kodiert. Dieser Vektore (das Positional Encoding) muss folgende Eigenschaften haben:

* **Eindeutig**: Jede Position muss ein einzigartiges Encoding erhalten.
* **Skalierbar**: Es sollte auch für unterschiedlich lange Sätze, die das Modell im Training noch nicht gesehen hat, funktionieren.
* **Sinnvolle Abstände**: Der Abstand zwischen den Encodings von zwei Positionen sollte unabhängig von der Satzlänge konsistent sein. Das ist wichtig, damit das Modell relative Positionen erlernen kann (z.B. _direkt daneben_ oder _drei Positionen entfernt_).
* **Addierbar zu Embeddings**: Es sollte sich gut mit den Wort-Embeddings kombinieren lassen, ohne deren Bedeutung zu überdecken (z.B. durch Additiond).
